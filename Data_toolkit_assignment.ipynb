2. How does broadcasting work in NumPy?
  - Broadcasting in NumPy is a powerful mechanism that allows NumPy to perform element-wise operations on arrays of different shapes without explicitly copying data.

  What is Broadcasting?

  Broadcasting automatically expands the smaller array so that it matches the shape of the larger array during an operation.

  This makes code:
  * Shorter
  * Faster
  * More memory-efficient

  import numpy as np

  a = np.array([1, 2, 3])
  print(a + 10)

  How it works:

  Scalar 10 is treated as [10, 10, 10]

  Result: [11, 12, 13]

4. Explain the use of the groupby() method in Pandas.
  - The groupby() method in Pandas is used to group data based on one or more columns and then perform aggregation or analysis on each group.

  It follows the classic “split-apply-combine” strategy.
  
  What does groupby() do?

  * Split the DataFrame into groups based on a column (or columns)
  * Apply a function (sum, mean, count, etc.) to each group
  * Combine the results into a new DataFrame or Series
  Basic Syntax :
  df.groupby("column_name").agg(function)

  import pandas as pd

  data = {
     "Department": ["IT", "HR", "IT", "HR", "Finance"],
     "Salary": [60000, 50000, 65000, 52000, 70000]
  }

  df = pd.DataFrame(data)
  df.groupby("Department")["Salary"].mean()

  Output : Department
           Finance    70000
           HR         51000
           IT         62500

8. What does the term “vectorized operation” mean in NumPy?
  - In NumPy, a vectorized operation means performing an operation on entire arrays at once, without using explicit Python loops.

  What does it mean?

  Instead of processing elements one by one, NumPy applies the operation element-wise to all values in the array simultaneously using optimized C code.

  import numpy as np

  a = np.array([1, 2, 3, 4])

  b = a * 2
  print(b)

  output: [2 4 6 8]

  Here, multiplication happens on the whole array in one step, not in a loop.

17. Explain the difference between apply() and map() in Pandas?
  - In Pandas, both apply() and map() are used to apply functions, but they differ in scope and use cases.

  map()

  * Works on Series only
  * Applies a function element-wise
  * Often used for value substitution or transformation
  * Can accept a function, dictionary, or Series

  Example:

  s.map(lambda x: x * 2)

  apply()
  * Works on Series and DataFrames
  * Applies a function along an axis (rows or columns)
  * More flexible and powerful than map()

  Example:

  df.apply(sum, axis=0)   # column-wise

21. Why is NumPy's array slicing faster than Python's list slicing?
  - NumPy's array slicing is faster than Python's list slicing because of how NumPy is designed internally.

  Why NumPy slicing is faster:
  * Views, not copies: NumPy slicing usually returns a view of the original array, while list slicing creates a new list
  * Contiguous memory storage: NumPy arrays are stored in continuous memory blocks, enabling faster access
  * Low-level (C-based) implementation: Operations are executed in optimized C code, not Python loops
  * Reduced overhead: No per-element Python object handling

```python
# 4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample DataFrame
data = {
    'Math': [85, 90, 78, 92],
    'Science': [88, 76, 91, 85],
    'English': [82, 89, 84, 90]
}
df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

# Visualize with heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix Heatmap")
plt.show()
```
